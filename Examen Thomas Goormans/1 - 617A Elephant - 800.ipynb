### 617A Elephant

https://codeforces.com/problemset/problem/617/A

```python
travelDistance = int(input())
stepsTaken = 0
for i in range(5, 0, -1):
    if travelDistance == 0:
        break
    else:
        stepsTaken += (travelDistance // i)
        travelDistance -= ((travelDistance // i)*i)
print(stepsTaken)
```

```python
travelDistance = int(input())
if travelDistance % 5 == 0:
    stepsTaken = (travelDistance // 5)
else:
    stepsTaken = (travelDistance // 5) + 1

print(stepsTaken)
```
