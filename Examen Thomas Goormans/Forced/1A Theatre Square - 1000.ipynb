### 1A Theatre Square

https://codeforces.com/problemset/problem/1/A

```python
dimensionsList = input().split(" ")
n = int(dimensionsList[0])
m = int(dimensionsList[1])
a = int(dimensionsList[2])
if n % a == 0:
    widthCount = (n//a)
else:
    widthCount = (n//a) + 1
if m % a == 0:
    lengthCount = (m//a)
else:
    lengthCount = (m//a) + 1
print(widthCount*lengthCount)
```
