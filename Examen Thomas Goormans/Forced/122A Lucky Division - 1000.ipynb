### 122A Lucky Division

https://codeforces.com/problemset/problem/122/A

```python
def luckyNumbers(x):
    result = []
    for i in range(1, x+1):
        luckyNumber = True
        iString = str(i)
        for j in iString:
            if j not in '47':
                luckyNumber = False
        if luckyNumber:
            result += [i]    
    return result

numberToCheck = int(input())
luckyNubersList = luckyNumbers(numberToCheck)
almostLucky = False
for i in luckyNubersList:
    if numberToCheck % i == 0:
        almostLucky = True
        break

if almostLucky:
    print("YES")
else:
    print("NO")
```
