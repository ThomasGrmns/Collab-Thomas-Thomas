### 71A Way Too Long words

https://codeforces.com/problemset/problem/71/A

```python
stringsCount = int(input().split()[0])
stringsList = []

for i in range(stringsCount):
    stringsList += input().split()

for string in stringsList:
    stringLength = len(string)
    if len(string) <= 10:
        print(string)
    else:
        print(f"{string[0]}{stringLength-2}{string[-1]}")
```
