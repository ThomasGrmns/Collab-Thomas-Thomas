### 236A Boy or Girl

https://codeforces.com/problemset/problem/236/A

```python
name = input()
uniqueLetters = 0
from string import ascii_lowercase as alc
for i in alc:
    for letter in name:
        if letter == i:
            uniqueLetters += 1
            break
if uniqueLetters % 2 == 0:
    print("CHAT WITH HER!")
else:
    print("IGNORE HIM!")
```
