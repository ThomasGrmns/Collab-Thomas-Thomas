### 96A Football

https://codeforces.com/problemset/problem/96/A

```python
playerPosition = input()
sameTeamCount = 1
situationDangerous = False
for p in range(1, len(playerPosition)):
    if playerPosition[p] == playerPosition[p-1]:
        sameTeamCount += 1
        if sameTeamCount == 7:
            situationDangerous = True
            break
    else:
        sameTeamCount = 1
            
if situationDangerous:
    print("YES")
else:
    print("NO")
```
