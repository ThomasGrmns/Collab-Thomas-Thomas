### 507B Amr and Pins

https://codeforces.com/problemset/problem/507/B

```python
r, x1, y1, x2, y2 = input().split(" ")
r = int(r)
x1 = int(x1)
y1 = int(y1)
x2 = int(x2)
y2 = int(y2)
d = (((x2 - x1)**2) + ((y2 - y1)**2)) ** 0.5
if d % (2*r) == 0:
    result = d // (2*r)
else:
    result = (d // (2*r)) + 1
print(int(result))
```
