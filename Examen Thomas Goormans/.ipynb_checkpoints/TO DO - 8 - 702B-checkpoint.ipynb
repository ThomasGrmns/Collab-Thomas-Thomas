```python
import math

n = int(input().split()[0])
numbers = list(map(int, input().split(" ")))
maxNumber = int(sorted(numbers)[-1])
powersOfTwo = []
x = 0

while (2**x) <= (maxNumber*2):
    powersOfTwo.append(2**x)
    x += 1

numberOfPairs = 0
for i in range(n-1):
    if numbers[i] % 2 == 0:
        for j in range((i+1), n):
            if numbers[j] % 2 == 0:
                pairsSum = numbers[i]+numbers[j]
                if pairsSum % 2 == 0:
                    if pairsSum in powersOfTwo:
                        numberOfPairs += 1
    else:
        for j in range((i+1), n):
            if numbers[j] % 2 != 0:
                pairsSum = numbers[i]+numbers[j]
                if pairsSum % 2 == 0:
                    if pairsSum in powersOfTwo:
                        numberOfPairs += 1
#print(numbers)
#print(powersOfTwo)
print(numberOfPairs)
```
