### 1996A Legs

https://codeforces.com/problemset/problem/1996/A

```python
testCasesCount = int(input().split()[0])

for i in range(testCasesCount):
    numberOfLegs = int(input().split()[0])
    numberOfAnimals = 0
    while numberOfLegs > 0:
        if numberOfLegs // 4 > 0:
            numberOfAnimals += (numberOfLegs // 4)
            numberOfLegs %= 4
        else:
            numberOfAnimals += (numberOfLegs // 2)
            numberOfLegs %= 2
    print(numberOfAnimals)
    
```
