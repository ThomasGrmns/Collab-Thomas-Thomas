### Ultra-Fast Mathematician

https://codeforces.com/problemset/problem/61/A

```python
firstNumber = input().split()[0]
secondNumber = input().split()[0]
resultNumber = ""
for i in range(len(firstNumber)):
    if firstNumber[i] == secondNumber[i]:
        resultNumber += "0"
    else:
        resultNumber += "1"
print(resultNumber)
```
