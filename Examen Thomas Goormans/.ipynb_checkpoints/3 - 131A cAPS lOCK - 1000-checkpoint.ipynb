### 131 cAPS lOCK

https://codeforces.com/problemset/problem/131/A

```python
myString = input()
allCapsCheck = myString.upper()
onlyFirstLower = myString.lower().capitalize().swapcase()

if myString == allCapsCheck:
    print(myString.swapcase())
elif myString == onlyFirstLower:
    print(myString.swapcase())
else:
    print(myString)
```
