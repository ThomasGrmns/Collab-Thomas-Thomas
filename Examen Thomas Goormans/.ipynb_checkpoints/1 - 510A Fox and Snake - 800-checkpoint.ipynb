### 510A Fox and Snake

https://codeforces.com/problemset/problem/510/A

```python
patternDimensions = input().split(" ")
n = int(patternDimensions[0])
m = int(patternDimensions[1])
for i in range(1, n+1):
    if i % 2 == 0:
        if i % 4 == 0:
            print(f"{'#'}{'.'*(m-1)}")
        else:
            print(f"{'.'*(m-1)}{'#'}")
    else:
        print('#'*m)
```
