### 349A - Cinema Line

https://codeforces.com/problemset/problem/349/A

```python
def listAdd(a, b):
    result = []
    for i in range(len(a)):
        result.append(a[i]+b[i])
    return result

def checkRegister(register):
    changePossible = True
    for i in range(3):
        if register[i] < 0:
            changePossible = False
            break
    return changePossible
    
customers = int(input().split(" ")[0])
cash = input().split(" ")
register = [0, 0, 0]
changePossible = True

for c in cash:
    if int(c) == 100:
        if register[1] > 0:
            register = listAdd(register, [-1, -1, 1])
            changePossible = checkRegister(register)
        else:
            register = listAdd(register, [-3, 0, 1])
            changePossible = checkRegister(register)
    elif int(c) == 50:
        register = listAdd(register, [-1, 1, 0])
        changePossible = checkRegister(register)
    else:
        register = listAdd(register, [1, 0, 0])
    if changePossible == False:
        break
if changePossible == True:
    print("YES")
else:
    print("NO")
```
