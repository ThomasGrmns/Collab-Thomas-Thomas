### 318A - Even odds

https://codeforces.com/problemset/problem/318/A

```python
n, p = input().split(" ")
n = int(n)
p = int(p)

if n % 2 == 0:
    if p > n/2:
        positionP = (p - (n/2))*2
    else:
        positionP = (p*2) - 1
else:
    if p > ((n//2)+1):
        positionP = (p-(n//2)-1)*2
    else:
        positionP = (p*2) - 1
print(int(positionP))
```
