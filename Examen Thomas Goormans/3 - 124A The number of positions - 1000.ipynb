### 124A - The number of positions

https://codeforces.com/problemset/problem/124/A

```python
n, a, b = input().split(" ")
totalLine = int(n)
beforePetr = int(a)
behindPetr = int(b)
possiblePositions = 0
for i in range(totalLine+1):
    if beforePetr < i >= (totalLine-behindPetr):
        possiblePositions += 1
print(possiblePositions)
```
