### 112A Petya and Strings

https://codeforces.com/problemset/problem/112/A

```python
firstString = input()
secondString = input()
stringLength = len(firstString)
isEqual = True
isSmaller = False
isGreater = False
for i in range(stringLength):
    if firstString[i].lower() < secondString[i].lower():
        isEqual = False
        isSmaller = True
        break
    elif firstString[i].lower() > secondString[i].lower():
        isEqual = False
        isGreater = True
        break
    else:
        continue

if isEqual:
    print(0)
elif isSmaller:
    print(-1)
else:
    print(1)
```
